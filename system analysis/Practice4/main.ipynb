```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, recall_score, precision_score, roc_auc_score

# 1. Загрузка данных
train_df = pd.read_csv('candy-data.csv')
test_df = pd.read_csv('candy-test.csv')

# 2. Признаки
features = [
    'chocolate', 'fruity', 'caramel', 'peanutyalmondy', 'nougat',
    'crispedricewafer', 'hard', 'bar', 'pluribus', 'sugarpercent', 'pricepercent'
]
target = 'Y'

# 3. Исключаем три конфеты из тренировочного набора
exclude = ['Milky Way', 'Now & Later', 'Nestle Smarties']
train_df = train_df[~train_df['competitorname'].isin(exclude)]

X_train = train_df[features]
y_train = train_df[target]

# 4. Обучение модели
model = LogisticRegression(random_state=2019, solver='lbfgs')
model.fit(X_train, y_train)

# 5. Предсказание вероятностей для тестового набора
proba_test = model.predict_proba(test_df[features])[:, 1]

# 6. Вероятности для конкретных конфет
sugar_babies = test_df[test_df['competitorname'] == 'Sugar Babies']
prob_sb = model.predict_proba(sugar_babies[features])[0, 1]
print(f"Sugar Babies: {prob_sb:.3f}")

tootsie_snack = test_df[test_df['competitorname'] == 'Tootsie Roll Snack Bars']
prob_ts = model.predict_proba(tootsie_snack[features])[0, 1]
print(f"Tootsie Roll Snack Bars: {prob_ts:.3f}")

# 7. Оценка модели на тестовом наборе (порог 0.5)
y_true = test_df['Y']
y_pred = (proba_test >= 0.5).astype(int)

print("\nМатрица ошибок:")
print(confusion_matrix(y_true, y_pred))

print(f"Recall (TPR): {recall_score(y_true, y_pred):.3f}")
print(f"Precision: {precision_score(y_true, y_pred):.3f}")
print(f"AUC: {roc_auc_score(y_true, proba_test):.3f}")
```
