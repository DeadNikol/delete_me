```python
import pandas as pd
import numpy as np
from scipy.spatial.distance import euclidean, cityblock

df = pd.read_csv("./tipa.csv")

# Новый объект
new_point = np.array([67, 36])

# --- ЕВКЛИДОВО РАССТОЯНИЕ ---
# Вычисляем расстояние от новой точки до каждой строки
df['Euclidean_Dist'] = df.apply(lambda row: euclidean([row['X'], row['Y']], new_point), axis=1)

# Сортируем по расстоянию и берем топ-3
top3_euclidean = df.sort_values(by='Euclidean_Dist').head(3)

print("=== ЕВКЛИДОВА МЕТРИКА ===")
print(f"Расстояние до ближайшего: {top3_euclidean['Euclidean_Dist'].iloc[0]:.3f}")
print(f"ID трёх ближайших: {', '.join(map(str, top3_euclidean['id'].tolist()))}")
# Класс определяется как самый частый (мода) среди соседей
print(f"Класс: {int(top3_euclidean['Class'].mode()[0])}")
print()


# --- МАНХЭТТЕНСКОЕ РАССТОЯНИЕ ---
# cityblock в scipy - это и есть манхэттенское расстояние
df['Manhattan_Dist'] = df.apply(lambda row: cityblock([row['X'], row['Y']], new_point), axis=1)

# Сортируем по расстоянию и берем топ-3
top3_manhattan = df.sort_values(by='Manhattan_Dist').head(3)

print("=== МАНХЭТТЕНСКАЯ МЕТРИКА ===")
print(f"Расстояние до ближайшего: {top3_manhattan['Manhattan_Dist'].iloc[0]:.3f}")
print(f"ID трёх ближайших: {', '.join(map(str, top3_manhattan['id'].tolist()))}")
print(f"Класс: {int(top3_manhattan['Class'].mode()[0])}")
```
