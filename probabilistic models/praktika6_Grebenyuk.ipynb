```python
import numpy as np

print("NumPy импортирован успешно")
```

```python
P = np.array([
    [0.8, 0.2],
    [0.4, 0.6],
])

k = P.shape[0]  # число состояний

# Строим систему уравнений (P^T - I)*pi^T = 0, с заменой
# последней строки на условие нормировки
A = np.vstack([(P.T - np.eye(k))[:-1], np.ones(k)])

b = np.zeros(k)
b[-1] = 1  # последнее уравнение соответствует условию суммы = 1

pi_exact = np.linalg.solve(A, b)  # решаем систему линейных уравнений

print("Стационарное распределение (точный метод):", pi_exact)
print("Ожидаемое значение: [0.6667, 0.3333]")
```

```python
# Итерационный метод
pi_iter = np.array([0.5, 0.5])  # начинаем с равномерного распределения

for _ in range(100):  # 100 итераций достаточно для хорошей точности
    pi_iter = pi_iter @ P

print("Стационарное распределение (итерационный метод):", pi_iter)

# Метод собственных векторов
eigvals, eigvecs = np.linalg.eig(P.T)  # обратите внимание: P.T, а не P!

idx = np.argmin(np.abs(eigvals - 1))  # ищем индекс собственного значения, равного 1

pi_eig = np.real(eigvecs[:, idx])
pi_eig = pi_eig / np.sum(pi_eig)  # нормировка, чтобы сумма была равна 1

print("Стационарное распределение (метод собственных векторов):", pi_eig)

print("\nСравнение всех трёх методов:")
print(f"  Точный (линейная система):      {pi_exact}")
print(f"  Итерационный:                   {pi_iter}")
print(f"  Собственные векторы:            {pi_eig}")
```

Поэтому для нахождения π как вектора-строки нужно брать собственный вектор Pᵀ, отвечающий λ = 1, и нормировать его так, чтобы сумма компонент равнялась 1.</td></tr>

```python
states3 = ['Главная', 'Каталог', 'Корзина']

P3 = np.array([
    [0.5, 0.4, 0.1],
    [0.3, 0.5, 0.2],
    [0.6, 0.1, 0.3],
])

k3 = P3.shape[0]

A3 = np.vstack([(P3.T - np.eye(k3))[:-1], np.ones(k3)])
b3 = np.zeros(k3)
b3[-1] = 1

pi3 = np.linalg.solve(A3, b3)

print("Стационарное распределение (3 состояния):", pi3)

# Проверка через высокую степень матрицы
P3_100 = np.linalg.matrix_power(P3, 100)  # 100 — достаточно для хорошей проверки

print("P^100 =")
print(P3_100)
print("(все строки должны быть близки к найденному pi3)")
```

```python
P_reducible = np.array([
    [1.0, 0.0],
    [0.0, 1.0],
])

# Проверим, что ЛЮБОЕ распределение (a, 1-a) стационарно
for a in [0.2, 0.5, 0.9]:
    pi_test = np.array([a, 1 - a])
    result = pi_test @ P_reducible
    print(f"pi=({a}, {1-a:.1f}) → pi·P = {result} (совпадает с pi? {np.allclose(result, pi_test)})")
```

```python
P_periodic = np.array([
    [0.0, 1.0],
    [1.0, 0.0],
])

print("Степени матрицы P_periodic:")
for n in [1, 2, 3, 4, 5, 6]:
    Pn = np.linalg.matrix_power(P_periodic, n)
    print(f"P^{n} =")
    print(Pn)

# Симуляция из состояния 0: смотрим, сходится ли распределение
current = 0
for step in range(6):
    print(f"Шаг {step}: состояние = {current}")
    current = np.random.choice([0, 1], p=P_periodic[current])
```
